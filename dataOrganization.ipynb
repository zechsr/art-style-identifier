```python
from datasets import load_dataset, ClassLabel, Features

ds = load_dataset("huggan/wikiart", split="train")

# 1. Get style names and map each label to a name
style_feat = ds.features["style"]
names = style_feat.names
print(names)  # check the exact cubism names

# 2. Merge cubism styles into one name
def merge_name(n):
    return "Cubism" if "cubism" in n.lower() else n

ds = ds.map(lambda ex: {"style_name": merge_name(names[ex["style"]])})

# 3. Remove styles under 1% of the data
from collections import Counter
counts = Counter(ds["style_name"])
total = len(ds)
keep = {s for s, c in counts.items() if c / total >= 0.01}
print("Dropped:", {s: c for s, c in counts.items() if s not in keep})

ds = ds.filter(lambda ex: ex["style_name"] in keep)

# 4. Re-encode as a ClassLabel with new contiguous ids
new_names = sorted(keep)
ds = ds.cast_column("style_name", ClassLabel(names=new_names))
ds = ds.remove_columns("style").rename_column("style_name", "style")

print(ds.features["style"])
print(Counter(new_names[i] for i in ds["style"]))
```
